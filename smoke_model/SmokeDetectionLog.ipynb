**Smoke detection version af `CropCalculatorLog.ipynb`.** Koden er den samme som originalen; kun data-stien og udvælgelsen af features/target er ændret. Datasæt: `data/smoke_detection_iot.csv`. `Unnamed: 0` (index), `UTC` (tidsstempel) og `CNT` (tæller) er droppet. Features = kun `Temperature[C]`, `Humidity[%]` og `PM2.5` (de tre sensorer i Forest Fire Sensor Network-projektet; alle numeriske, så `cat_attribs` er tom), target = `Fire Alarm`. Split stratificeres på `Fire Alarm` i stedet for `Entity`. `np.log1p` bruges på `PM2.5` (meget skæv fordeling) i stedet for `Nitrogen_fertilizer_use`. ("Log" = log-transformationen; modellen er RandomForestRegressor som i originalen.) `max_features`-griddet er ændret fra [4, 6, 8] til [1, 2, 3], da der kun er 3 features.

**Note:** `Fire Alarm` er binær (0/1), men modellen er stadig en *regressor* ligesom i originalen. Forudsigelserne er derfor et tal (ca. mellem 0 og 1), og RMSE/MSE/R2/MAE måler hvor tæt det tal er på 0/1. Det er ikke en klassifikations-accuracy.

```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np
import plotly.express as px
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import SGDRegressor


from sklearn.preprocessing import FunctionTransformer
import numpy as np

```

```python
#df=pd.read_csv("data/cerealCrop.csv")
df=pd.read_csv("data/smoke_detection_iot.csv")
df = df.drop(columns=["Unnamed: 0", "UTC", "CNT"]) # index, tidsstempel og tæller er ikke features
df.head()

```

```python
train_clean = train_set.dropna(subset=["Fire Alarm"])

X_train = train_clean.drop("Fire Alarm", axis=1)

y_train = train_clean["Fire Alarm"].copy()


num_attribs = ["Temperature[C]", "Humidity[%]", "PM2.5"]

cat_attribs = [] # ingen kategoriske kolonner i dette datasæt



num_pipeline = make_pipeline(
 SimpleImputer(strategy="median"),

 StandardScaler(),
)

cat_pipeline = make_pipeline(
 SimpleImputer(strategy="most_frequent"),
 OneHotEncoder(handle_unknown="ignore")) 


preprocessing = ColumnTransformer([
 ("num", num_pipeline, num_attribs),
 ("cat", cat_pipeline, cat_attribs),
])

X_train_prepared = preprocessing.fit_transform(X_train) # manglede i originalen
print(X_train_prepared.shape)


```

```python
#tree_reg = make_pipeline(preprocessing, LinearRegression())
#tree_reg.fit(X_valueFromMatrix, Yield_labels)
tree_reg = make_pipeline(preprocessing, RandomForestRegressor(random_state=42))
tree_reg.fit(X_train, y_train)

#tree_reg = make_pipeline(preprocessing,SGDRegressor(max_iter=1000, tol=1e-3, penalty=None, eta0=0.00001, random_state=42))
#tree_reg.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

test_clean = test_set.dropna(subset=["Fire Alarm"])
X_test = test_clean.drop("Fire Alarm", axis=1)
y_test = test_clean["Fire Alarm"].copy()

yield_predictions = tree_reg.predict(X_test)

tree_mse = mean_squared_error(y_test, yield_predictions)
tree_rmse = np.sqrt(tree_mse)

print(f"Din endelige test-MSE er: {tree_mse:.4f}")
print(f"Modellens gennemsnitlige fejl før finetuning (RMSE): {tree_rmse}")
final_r2 = r2_score(y_test, yield_predictions)
print(f"Din endelige test-R2 score er: {final_r2:.4f}")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline


full_pipeline = Pipeline([
 ("preprocessing", preprocessing),
 ("random_forest", RandomForestRegressor(random_state=42)),
])
param_grid = [
    {
        'random_forest__n_estimators': [30, 100, 150], 
        'random_forest__max_features': [1, 2, 3], # [4, 6, 8] i originalen - der er kun 3 features nu
        'random_forest__max_depth': [None, 10, 20],
        'random_forest__min_samples_leaf': [1, 2, 4]

    },
]
grid_search = GridSearchCV(full_pipeline, param_grid, cv=10,
 scoring='neg_root_mean_squared_error', n_jobs=-1) # n_jobs=-1 kun for hastighed
grid_search.fit(X_train, y_train)


cv_res = pd.DataFrame(grid_search.cv_results_)
cv_res.sort_values(by="mean_test_score", ascending=False, inplace=True)

```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

test_clean = test_set.dropna(subset=["Fire Alarm"])

X_test = test_clean.drop("Fire Alarm", axis=1)
y_test = test_clean["Fire Alarm"].copy()

final_model = grid_search.best_estimator_


final_predictions = final_model.predict(X_test)

final_mse = mean_squared_error(y_test, final_predictions)
final_rmse = np.sqrt(final_mse)

print("random forest finetuning resultater:")
print(f"Din endelige test-MSE er: {final_mse:.4f}")
print(f"Din endelige test-RMSE er: {final_rmse:.4f}")
final_r2 = r2_score(y_test, final_predictions)
print(f"Din endelige test-R2 score er: {final_r2:.4f}")
final_mae = mean_absolute_error(y_test, final_predictions)
print(f"Din endelige test-MAE er: {final_mae:.4f}")
```

```python
from sklearn.model_selection import learning_curve
import matplotlib.pyplot as plt
import numpy as np

# Brug den beste model fra grid search
best_model = grid_search.best_estimator_

# Generer learning curves
train_sizes, train_scores, val_scores = learning_curve(
    best_model,
    X_train,
    y_train,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    train_sizes=np.linspace(0.1, 1.0, 10),
    shuffle=True,
    random_state=42
)

# Beregn gennemsnit og std
train_mean = -train_scores.mean(axis=1)  # Negativ fordi scoring er neg_rmse
train_std = train_scores.std(axis=1)
val_mean = -val_scores.mean(axis=1)
val_std = val_scores.std(axis=1)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(train_sizes, train_mean, label="Training RMSE", marker='o')
plt.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.2)
plt.plot(train_sizes, val_mean, label="Validation RMSE", marker='s')
plt.fill_between(train_sizes, val_mean - val_std, val_mean + val_std, alpha=0.2)
plt.xlabel("Training Set Size")
plt.ylabel("RMSE")
plt.title("Learning Curve")
plt.legend()
plt.grid(True)
plt.show()

# Diagnostik
print(f"Training RMSE (final): {train_mean[-1]:.4f}")
print(f"Validation RMSE (final): {val_mean[-1]:.4f}")
print(f"Gap: {(val_mean[-1] - train_mean[-1]):.4f}")

if val_mean[-1] - train_mean[-1] > 0.15:
    print("⚠️ Tegn på OVERFITTING - validation fejl meget højere end training")
elif train_mean[-1] > 1.0:
    print("⚠️ Tegn på UNDERFITTING - begge fejl høje")
else:
    print("✅ God balance - modellen generaliserer godt")
```
